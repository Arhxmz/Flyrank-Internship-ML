Search Question: Which published client pages are experiencing significant organic traffic decline and should be prioritized for a content refresh this month?
Unit of Analysis: A single published client page (represented by a page snapshot/ID).
Output:** A risk score or binary flag (`is_declining_label`) identifying whether a page is in a sustained downward trend.
The Decision:** Determining which top 10%–20% of decaying pages should be queued for editorial review versus which stable/growing pages should be left untouched.
The Action:** An SEO editor reviews the flagged page, updates outdated facts, improves content depth, and re-optimizes on-page elements.
Cost of a Wrong Call:**
False Positive (Flagging a healthy page):** Wasted editorial budget (~$150–$300 per article update) spent revising content that did not need changes, diverting resources away from actual problem pages.
False Negative (Missing a declining page):** Continued organic rank decay, resulting in unmitigated loss of search impressions, organic visits, and client conversions over several months.


## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the starter dataset (uploaded to Colab root folder)
df = pd.read_csv("content_refresh_anonymized.csv")

# Extract baseline numbers
total_pages = len(df)
total_clients = df['client_id'].nunique() if 'client_id' in df.columns else "N/A"

# Trend direction distribution
decline_pct = (df['trend_direction'] == 'down').mean() * 100

# Calculate clicks at risk
click_col = 'clicks_last_30d' if 'clicks_last_30d' in df.columns else ('clicks_30d' if 'clicks_30d' in df.columns else None)
total_clicks_at_risk = df[df['trend_direction'] == 'down'][click_col].sum() if click_col else "N/A"

print("=== SECTION 3: BASELINE DATA SUMMARY ===")
print(f"1. Total Pages Analyzed: {total_pages:,}")
print(f"2. Unique Client Domains: {total_clients}")
print(f"3. Pages in Downward Trend: {decline_pct:.1f}% of all pages")
if click_col:
    print(f"4. Total Monthly Clicks at Risk: {total_clicks_at_risk:,}")
```

## 4. Careful words: what I can and can't claim

### 4. Trade-offs, Edge Cases, and Decision-Support Language

* **Trade-offs:** Prioritizing pages purely by traffic decline risks wasting editorial resources on seasonal drops or pages where search intent has permanently shifted away from our offer. Conversely, strictly prioritizing high-traffic pages might ignore lower-traffic, high-conversion pages.
* **Edge Cases:** Brand new pages without sufficient historical baseline data; pages experiencing temporary site-wide migration issues; highly seasonal content (e.g., holiday guides) showing expected cyclical traffic drops.
* **Decision-Support Language:** Predictions produced by this workflow should be framed as directional guidance (e.g., *"Page X is measured to be in a strong downward trend and is recommended for review"*) rather than definitive business mandates, allowing human editors to retain final decision authority.
